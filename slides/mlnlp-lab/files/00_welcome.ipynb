- 能用 `pandas` 快速做基础统计和筛选。

```python
import pandas as pd
import matplotlib.pyplot as plt

print('pandas 版本：', pd.__version__)
print('matplotlib 版本：', plt.matplotlib.__version__)
```

## 1）读取数据
先把课程自带的 `data/text_samples.csv` 读进来。


```python
df = pd.read_csv('data/text_samples.csv')
df
```

## 2）理解数据结构
看清行列、字段类型、缺失值。


```python
print('shape:', df.shape)
print('columns:', list(df.columns))
print('\n每列类型:')
print(df.dtypes)

print('\n缺失值:')
print(df.isna().sum())
```

## 3）分类分布统计
统计每个 `label` 的样本数，快速理解数据平衡性。


```python
label_counts = df['label'].value_counts()
print(label_counts)

label_counts.plot(kind='bar', title='label 分布')
plt.ylabel('样本数')
plt.tight_layout()
plt.show()
```

## 4）特征工程（入门级）
给每条文本加一个 `text_length` 特征，后续分析更直观。


```python
df['text'] = df['text'].astype(str).str.strip()
df['text_length'] = df['text'].str.len()

df[['id', 'text', 'label', 'text_length']]
```

## 5）按标签分组做平均长度


```python
length_mean = (
    df.groupby('label', as_index=False)['text_length']
    .mean()
    .sort_values('text_length', ascending=False)
)
length_mean

length_mean.plot(x='label', y='text_length', kind='bar', legend=False, rot=0)
plt.ylabel('平均文本长度')
plt.title('不同标签的平均文本长度')
plt.tight_layout()
plt.show()
```

## 6）课堂小练习（10 分钟）
1. 只筛选 `positive` 标签。
2. 在 positive 里找出长度大于 10 的文本。
3. 看看 neutral 和 negative 分别有哪些文本。


```python
positive_df = df[df['label'] == 'positive']
print('positive 数量:', len(positive_df))
print(positive_df)

positive_long = positive_df[positive_df['text_length'] > 10]
print('positive 且 text_length > 10:')
print(positive_long)

for target in ['neutral', 'negative']:
    print('\n====', target, '====')
    print(df[df['label'] == target])
```
