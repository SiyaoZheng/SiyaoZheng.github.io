## 示例数据

下面读取一个课程自带的小型中文文本数据集，做一个最简单的统计和可视化。

````python
import matplotlib.pyplot as plt
import pandas as pd

df = pd.read_csv("data/text_samples.csv")
df["text_length"] = df["text"].str.len()
df
````

````python
summary = df.groupby("label", as_index=False)["text_length"].mean()

ax = summary.plot(
    x="label",
    y="text_length",
    kind="bar",
    legend=False,
    color=["#2563eb", "#f97316", "#16a34a"]
)
ax.set_title("Average text length by label")
ax.set_xlabel("label")
ax.set_ylabel("average number of characters")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
````
